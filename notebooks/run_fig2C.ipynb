```python
import torch

# --- Make sure repo root is on PYTHONPATH so `import foil_project` works ---
import sys
from pathlib import Path

def add_repo_root_to_path(package_name: str = "foil_project") -> str:
    cwd = Path.cwd().resolve()
    for parent in [cwd, *cwd.parents]:
        pkg = parent / package_name
        if pkg.is_dir() and (pkg / "__init__.py").exists():
            sys.path.insert(0, str(parent))
            return str(parent)
    raise ModuleNotFoundError(
        f"Could not find '{package_name}' package folder above {cwd}. "
        f"Ensure your repo contains '{package_name}/__init__.py'."
    )

REPO_ROOT = add_repo_root_to_path("foil_project")
print("Repo root added to sys.path:", REPO_ROOT)

from foil_project.config import PathsConfig, PermutedMNISTConfig, TrainingConfig, ModelConfig
from foil_project.experiments import run_permuted_experiment_2C
from foil_project.plotting import plot_fig2c_paper_like
from foil_project.utils import get_device

device = get_device()
print("Device:", device)
```

```python
paths = PathsConfig()
ds_cfg = PermutedMNISTConfig()
tr_cfg = TrainingConfig()
mdl_cfg = ModelConfig()

results_2c = run_permuted_experiment_2C(
    paths=paths,
    ds_cfg=ds_cfg,
    tr_cfg=tr_cfg,
    mdl_cfg=mdl_cfg,
    device=device,
)

print("Input dims:", results_2c["input_dims"])
print("Layer names:", results_2c["layer_names"])
print("Overlap Fisher (6x2):")
for i, (name, overlap) in enumerate(zip(results_2c["layer_names"], results_2c["overlap_fisher"])):
    print(f"  Layer {i+1} ({name}): A vs B = {overlap[0]:.4f}, mean(A vs C, B vs C) = {overlap[1]:.4f}")
```
