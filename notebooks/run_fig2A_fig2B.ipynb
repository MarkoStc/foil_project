```python
import torch

# --- Make sure repo root is on PYTHONPATH so `import foil_project` works ---
import sys
from pathlib import Path

def add_repo_root_to_path(package_name: str = "foil_project") -> str:
    cwd = Path.cwd().resolve()
    for parent in [cwd, *cwd.parents]:
        pkg = parent / package_name
        if pkg.is_dir() and (pkg / "__init__.py").exists():
            sys.path.insert(0, str(parent))
            return str(parent)
    raise ModuleNotFoundError(
        f"Could not find '{package_name}' package folder above {cwd}. "
        f"Ensure your repo contains '{package_name}/__init__.py'."
    )

REPO_ROOT = add_repo_root_to_path("foil_project")
print("Repo root added to sys.path:", REPO_ROOT)


from foil_project.config import PathsConfig, PermutedMNISTConfig, RotatedMNISTConfig, TrainingConfig, ModelConfig
from foil_project.experiments import (
    run_full_permuted_experiment,
    run_fig2a_dynamics_permuted_abc,
    run_full_rotated_experiment,
    run_fig2a_dynamics_rotated_abc,
)
from foil_project.plotting import plot_fig2a_paper_like, plot_fig2b_paper_like
from foil_project.utils import get_device

# Toggle: compute / plot Naive L2 as well
COMPUTE_L2 = False

device = get_device()
print("Device:", device)
```

```python
paths = PathsConfig()
ds_cfg = PermutedMNISTConfig()
tr_cfg = TrainingConfig()
mdl_cfg = ModelConfig()

full_results = run_full_permuted_experiment(
    compute_l2=COMPUTE_L2,
    paths=paths,
    ds_cfg=ds_cfg,
    tr_cfg=tr_cfg,
    mdl_cfg=mdl_cfg,
    device=device,
)

print("Single-task baseline:", full_results["single_task_acc"])
print("Final avg acc (SGD):", full_results["sgd"]["avg_acc_curve"][-1])
print("Final avg acc (EWC):", full_results["ewc"]["avg_acc_curve"][-1])
if COMPUTE_L2 and full_results["l2"] is not None:
    print("Final avg acc (L2):", full_results["l2"]["avg_acc_curve"][-1])
```

```python
plot_fig2b_paper_like(full_results, compute_l2=COMPUTE_L2)
```

```python
dyn_results = run_fig2a_dynamics_permuted_abc(
    compute_l2=COMPUTE_L2,
    paths=paths,
    ds_cfg=ds_cfg,
    tr_cfg=tr_cfg,
    mdl_cfg=mdl_cfg,
    device=device,
)
```

```python
plot_fig2a_paper_like(dyn_results)
```

```python
# --- RotatedMNIST: Full 10-task experiment (for Fig 2B) ---
rot_ds_cfg = RotatedMNISTConfig()  # 0,10,...,90 degrees by default

rot_full_results = run_full_rotated_experiment(
    compute_l2=COMPUTE_L2,
    paths=paths,
    ds_cfg=rot_ds_cfg,
    tr_cfg=tr_cfg,
    mdl_cfg=mdl_cfg,
    device=device,
)

print("Rotated single-task baseline:", rot_full_results["single_task_acc"])
print("Rotated final avg acc (SGD):", rot_full_results["sgd"]["avg_acc_curve"][-1])
print("Rotated final avg acc (EWC):", rot_full_results["ewc"]["avg_acc_curve"][-1])
if COMPUTE_L2 and rot_full_results["l2"] is not None:
    print("Rotated final avg acc (L2):", rot_full_results["l2"]["avg_acc_curve"][-1])
```

```python
plot_fig2b_paper_like(rot_full_results, compute_l2=COMPUTE_L2, title="RotatedMNIST — Figure 2B-style")
```

```python
# --- RotatedMNIST: 3-task dynamics (A,B,C) (for Fig 2A) ---
rot_dyn_results = run_fig2a_dynamics_rotated_abc(
    compute_l2=COMPUTE_L2,
    paths=paths,
    ds_cfg=rot_ds_cfg,
    tr_cfg=tr_cfg,
    mdl_cfg=mdl_cfg,
    device=device,
)
```

```python
plot_fig2a_paper_like(rot_dyn_results, title="RotatedMNIST — Figure 2A-style (Tasks A,B,C)")
```
